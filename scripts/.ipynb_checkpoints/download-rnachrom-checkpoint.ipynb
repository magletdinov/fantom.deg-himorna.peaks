```python
# Все .txt файлы (аналог glob)
tar_gz_files = [i for i in Path(WORK_DIR).glob("*.tar.gz")]
```

```python
for to_tar_gz in tqdm(tar_gz_files):
    extract_to = Path(f"{to_tar_gz.stem.split(".tar")[0]}/")
    with tarfile.open(to_tar_gz, "r:gz") as tar:
        tar.extractall(path=extract_to)
    to_tar_gz.unlink()
```

```python
[i.unlink() for i in Path(tar_gz_files).glob("*index*")]
```
